```python
# "Relu" stands for rectified linear unit
# Example of element wise operation

# The main purpose of the element wise function is to apply a function to each element within the array independently to avoid slow loops
def naive_relu(x):
  assert len(x.shape) == 2
  x = x.copy()
  for i in range(x.shape[0]):
    for j in range(x.shape[1]):
      x[i, j] = max(x[i, j], 0)
```

```python
# Example of addition
def naive_add(x, y):
  assert len(x.shape) == 2
  assert x.shape == y.shape
  x = x.copy()
  for i in range(x.shape[0]):
    for j in range(x.shape[1]):
      x[i, j] += y[i, j]
  return x
```

Numpy has well built in functions in order to do the "heavy" lifting known as BLAS are low-level and highly parallel tensor manipulation routines implemented in Fortran and C

```python
import numpy as np
import time

x = np.random.random((20, 100))
y = np.random.random((20, 100))
t0 = time.time()
for _ in range(1000):
  z = x + y
  z = np.maximum(z, 0.)
print("Took: {0:.2f} s".format(time.time() - t0))
```
